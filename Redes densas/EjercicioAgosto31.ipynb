```python
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers, models, optimizers, losses, metrics

# Ruta robusta: intenta varias ubicaciones posibles
ruta_base = Path.cwd()
candidatos = [
    ruta_base / "Redes densas" / "pantalon.csv",
    ruta_base / "Apuntes-Ciencia-de-Datos" / "Redes densas" / "pantalon.csv",
    Path(r"C:\Users\ANGELA\Documents\Apuntes Ciencia de Datos\Apuntes-Ciencia-de-Datos\Redes densas\pantalon.csv")
]

ruta = next((p for p in candidatos if p.exists()), candidatos[0])
print("Ruta usada:", ruta)

# Cargar CSV como arreglo 2D
imagen = np.genfromtxt(ruta, delimiter=';', dtype=np.float32, filling_values=0)

# Asegurar que sea matriz 2D
imagen = np.asarray(imagen)
if imagen.ndim == 1:
    imagen = imagen.reshape(1, -1)

# Tamaño fijo de la figura
fig, ax = plt.subplots(figsize=(6, 6))

# Mostrar imagen con coordenadas
ax.imshow(
    imagen,
    cmap='gray',
    origin='upper',
    extent=[0, imagen.shape[1], 0, imagen.shape[0]]
)

# Coordenadas y etiquetas
ax.set_title("Imagen cargada desde CSV")
ax.set_xlabel("Coordenada X")
ax.set_ylabel("Coordenada Y")
ax.set_xlim(0, imagen.shape[1])
ax.set_ylim(imagen.shape[0], 0)  # invertido para que se vea como imagen

plt.show()
```
